```python
import os
base = '/kaggle/input'
print('input exists:', os.path.isdir(base))
def walk(path, depth=0, max_depth=3):
    try:
        names = sorted(os.listdir(path))
    except Exception as e:
        print('  ' * depth + f'<err {e}>')
        return
    if not names:
        print('  ' * depth + '(EMPTY)')
        return
    for n in names:
        p = os.path.join(path, n)
        isdir = os.path.isdir(p)
        size = '' if isdir else f' {os.path.getsize(p):,}B'
        print('  ' * depth + n + ('/' if isdir else '') + size)
        if isdir and depth < max_depth:
            walk(p, depth + 1, max_depth)
walk(base)
```
